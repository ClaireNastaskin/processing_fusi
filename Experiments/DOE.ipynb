This notebook is to organize a strategy for recording data in rats to determine the effects of gain and aperture.

```python
import itertools
import random

freq = ['f1f2'] # two frequencies, e.g. 6 and 7.5 MHz
aper = ['a1a2', 'a1a3','a2a3'] # three transmit apertures e.g. 16, 32, 64
gain = ['g1g3', 'g2g4'] # 4 gains, organized in 2 groups of 2 receive gains e.g. -1 and 1, 0 and 4

combinations = list(itertools.product(freq, aper, gain))
print(combinations)
# Label each combination with a run number
```

```python
# Note that ('f1f2', 'a1a2', 'g1g3') would correspond to 8 sequences for the pairwise combinations of the different levels of the three factors, as would be produced by the bash script + the baseline sequence. e.g.
freq = ['f1', 'f2']
aper = ['a1', 'a2']
gain = ['g1',  'g3']



combinations = list(itertools.product(freq, aper, gain))

# Add baseline sequence
baseline_sequence = ('f1', 'a2', 'g2')
combinations.append(baseline_sequence)

# Display all sequences including the baseline
for i, combination in enumerate(combinations):
    if i == len(combinations) - 1:
        print(f"Baseline Sequence: {combination}")
    else:
        print(f"Sequence {i+1}: {combination}")

```

Thus Combinations is a set of 6 runs that would be performed. Each would correspond to a sequence.yaml file. We would want to run all 6 runs at a single location in the rat.

The goal would be to record a total of 4 repetitions of the 6 runs. 2 repetitions within a rat at two distinct locations. 2 repetitions within a second rat at two distinct locations.

For each repetition we would want to randomize the order... so 


```python
labeled_combinations = [(i+1, comb) for i, comb in enumerate(combinations)]

# Generate 4 random orderings of the combinations
for repetition in range(4):
    random.shuffle(labeled_combinations)
    print(f"Random ordering {repetition + 1}:")
    for run_number, combination in labeled_combinations:
        print(f"Run {run_number}: {combination}")

```

Below is an alternative strategy we will not employ

```python
import itertools
import random

freq = ['f1', 'f2']
aper = ['a1', 'a2', 'a3']
gain = ['g1', 'g2', 'g3', 'g4']

combinations = list(itertools.product(freq, aper, gain))


# Shuffle the combinations list
random.shuffle(combinations)

# Divide into 3 equal groups
group_size = len(combinations) // 3
group1 = combinations[:group_size]
group2 = combinations[group_size:2*group_size]
group3 = combinations[2*group_size:]


print(group1)
print(group2)
print(group3)
```
